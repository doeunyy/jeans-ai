# Merge Data

whisper 모델을 통과시킨 결과를 담은 CSV 파일과 annotation 추출 csv 파일 병합

### Training 데이터셋 병합

```python
import pandas as pd

# CSV 파일 경로 설정
file1_path = "./whisper_train.csv"
file2_path = "./annotation_train.csv"
output_path = "./dataset_train.csv"  # 저장할 파일 경로

# CSV 파일 불러오기
df1 = pd.read_csv(file1_path)
df2 = pd.read_csv(file2_path)

# id 컬럼을 기준으로 병합 (첫 번째 파일의 컬럼이 먼저 오도록)
merged_df = df1.merge(df2, on="id", how="inner")  # inner join (공통 id만 유지)

# 병합된 데이터를 CSV 파일로 저장 (인덱스 없이 저장)
merged_df.to_csv(output_path, index=False, encoding="utf-8-sig")

print(f"병합된 파일이 '{output_path}' 경로에 저장되었습니다.")
```
