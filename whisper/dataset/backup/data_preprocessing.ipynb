### 데이터 압축 해제

```python
import os
import zipfile

def extract_zip_files(folder_path):
    """
    주어진 폴더 내의 모든 ZIP 파일을 압축 해제하는 함수

    Parameters:
    folder_path (str): ZIP 파일들이 위치한 폴더 경로
    """
    # ZIP 파일명을 저장할 리스트
    zip_file_names = []

    # ZIP 파일 순회
    for file_name in os.listdir(folder_path):
        if file_name.endswith(".zip"):
            zip_file_names.append(file_name)  # 파일명 추가
            zip_file_path = os.path.join(folder_path, file_name)

            # 압축 해제할 폴더 경로 (ZIP 파일명과 동일한 폴더 생성)
            extract_dir_name = os.path.splitext(file_name)[0]  # 확장자 제거한 이름름
            extract_dir_path = os.path.join(folder_path, extract_dir_name)
            os.makedirs(extract_dir_path, exist_ok=True)  # 디렉토리 생성

            # ZIP 파일 압축 해제
            with zipfile.ZipFile(zip_file_path, "r") as zip_ref:
                zip_ref.extractall(extract_dir_path)

            print(f"✅ 압축 해제 완료: {file_name} -> {extract_dir_path}")

    # ZIP 파일명 리스트 줄바꿈 출력
    print(f"\n📂 {folder_path} 내 처리된 ZIP 파일 리스트:")
    for zip_file in zip_file_names:
        print(zip_file)
```

```python
import re
import csv
import pandas as pd
import torch
import whisper
from pathlib import Path
```

```python
# whisper 모델 로드 함수

def load_whisper_model():
    device = "cuda" if torch.cuda.is_available() else "cpu"
    model = whisper.load_model("small").to(device)
    return model, device
```

```python
# 음성 파일 처리 함수

def process_audio_file(model, device, file_path):
    """
    Whisper 모델을 사용하여 오디오 파일을 텍스트로 변환
    """
    
    # 음성 파일 전처리: 음성 파일 로드 및 정규화
    audio = whisper.load_audio(file_path)
    audio = whisper.pad_or_trim(audio)
    
    mel = whisper.log_mel_spectrogram(audio).to(device)
    
    options = whisper.DecodingOptions(language="ko")
    result = whisper.decode(model, mel, options)
    
    return result.text
```

```python
def get_dir_contents(base_path: str, type: str = None):
    """
    디렉토리 내 폴더 및 파일 탐색
    """
    
    if type == "raw":
        raw_folders = [f for f in os.listdir(base_path) if f.startswith("[원천]") and os.path.isdir(os.path.join(base_path, f))]
        return raw_folders
    elif type == "wav":
        wav_files = [f for f in os.listdir(base_path) if f.endswith(".wav")]
        return wav_files
    elif type is None:
        all_contents = os.listdir(base_path)
        return all_contents
    else:
        raise ValueError("Invalid type. Use 'raw', 'wav', or leave it empty.")

# # test
# folders_raw = get_dir_contents(train_path, "raw")
# print(folders, len(folders_raw))
# subfolders = get_dir_contents(f"{train_path}/[원천]1.AI비서_원천_8_명령어(노인)_training", None)
# print(subfolders[:5])
# wav_files = get_dir_contents(f"{train_path}/[원천]1.AI비서_원천_8_명령어(노인)_training/script1_p_0089", "wav")
# print(len(wav_files))
```

```python
def process_data(base_path, folder_list, csv_file_path, model, device):
    """
    특정 폴더 목록을 순회하며 .wav 파일을 Whisper 모델에 통과시키고, 
    처리된 결과를 즉시 CSV 파일에 저장.

    Parameters:
        - base_path: 데이터 폴더의 기본 경로
        - folder_list: 처리할 폴더 목록
        - csv_file_path: 결과를 저장할 CSV 파일 경로
        - model: Whisper 모델 객체
        - device: 모델이 실행될 디바이스 (예: "cuda" 또는 "cpu")
    """

    # 파일이 존재하는지 확인
    file_exists = os.path.exists(csv_file_path)

    # CSV 파일이 없으면 헤더를 추가하여 생성
    if not file_exists:
        with open(csv_file_path, mode='w', encoding='utf-8-sig', newline='') as csv_file:
            writer = csv.DictWriter(csv_file, fieldnames=["id", "predText"])
            writer.writeheader()  # 헤더 추가

    # 폴더 리스트 순회
    for folder_idx, folder in enumerate(folder_list, start=1):
        match = re.search(r'_(\d+)_', folder)
        n_folder = match.group(1) if match else "Unknown"
        folder_path = os.path.join(base_path, folder)

        # 폴더 내 서브폴더 목록 가져오기
        subfolders = get_dir_contents(folder_path, None)

        for subfolder_idx, subfolder in enumerate(subfolders, start=1):
            subfolder_path = os.path.join(folder_path, subfolder)

            # 해당 서브폴더 안에 있는 .wav 파일 가져오기
            wav_files = get_dir_contents(subfolder_path, "wav")

            for file_idx, file in enumerate(wav_files, start=1):
                file_path = os.path.join(subfolder_path, file)
                file_id = Path(file).stem  # 확장자 제거한 파일명
                
                pred_text = process_audio_file(model, device, file_path)

                # CSV 파일에 즉시 저장
                with open(csv_file_path, mode='a', encoding='utf-8-sig', newline='') as csv_file:
                    writer = csv.DictWriter(csv_file, fieldnames=["id", "predText"])
                    writer.writerow({"id": file_id, "predText": pred_text})

                print(f"[{folder_idx}/{len(folder_list)}] Processing Folder {n_folder} 📂 {subfolder_idx}/{len(subfolders)} - 🎙️ {file_idx}/{len(wav_files)} - file_id: {file_id}")

    print(f"Processing complete. Results saved to {csv_file_path}")

```

### 중간에 실행 중단 시

```python
import re
import csv
import torch
import whisper
from pathlib import Path
```

```python
### 실행 중지 후 다시 실행시키고 싶다면?

def get_existing_file_ids(csv_file_path):
    """이미 처리된 파일 ID 목록을 가져옴"""
    if not os.path.exists(csv_file_path):
        return set()
    
    with open(csv_file_path, mode='r', encoding='utf-8-sig') as csv_file:
        reader = csv.DictReader(csv_file)
        return {row["id"] for row in reader}

def process_data_resumable(base_path, folder_list, csv_file_path, model, device):
    """
    특정 폴더 목록을 순회하며 .wav 파일을 Whisper 모델에 통과시키고, 
    이미 처리된 파일은 건너뛰어 중단된 부분부터 이어서 실행.
    """

    # 기존 CSV 파일에서 처리된 파일 ID 목록 가져오기
    processed_file_ids = get_existing_file_ids(csv_file_path)

    # CSV 파일 존재 여부 확인 (없으면 새로 생성)
    file_exists = os.path.exists(csv_file_path)
    if not file_exists:
        with open(csv_file_path, mode='w', encoding='utf-8-sig', newline='') as csv_file:
            writer = csv.DictWriter(csv_file, fieldnames=["id", "predText"])
            writer.writeheader()

    # 폴더 리스트 순회
    for folder_idx, folder in enumerate(folder_list, start=1):
        match = re.search(r'_(\d+)_', folder)
        n_folder = match.group(1) if match else "Unknown"
        folder_path = os.path.join(base_path, folder)

        # 폴더 내 서브폴더 목록 가져오기
        subfolders = get_dir_contents(folder_path, None)

        for subfolder_idx, subfolder in enumerate(subfolders, start=1):
            subfolder_path = os.path.join(folder_path, subfolder)

            # 해당 서브폴더 안에 있는 .wav 파일 가져오기
            wav_files = get_dir_contents(subfolder_path, "wav")

            for file_idx, file in enumerate(wav_files, start=1):
                file_path = os.path.join(subfolder_path, file)
                file_id = Path(file).stem  # 확장자 제거한 파일명

                # 이미 처리된 파일이면 건너뛰기
                if file_id in processed_file_ids:
                    print(f"🔄 Skipping already processed file: {file_id}")
                    continue

                # Whisper 모델을 사용하여 음성 인식 수행
                pred_text = process_audio_file(model, device, file_path)

                # 결과 CSV 파일에 즉시 저장
                with open(csv_file_path, mode='a', encoding='utf-8-sig', newline='') as csv_file:
                    writer = csv.DictWriter(csv_file, fieldnames=["id", "predText"])
                    writer.writerow({"id": file_id, "predText": pred_text})

                print(f"[{folder_idx}/{len(folder_list)}] Processing Folder {n_folder} 📂 {subfolder_idx}/{len(subfolders)} - 🎙️ {file_idx}/{len(wav_files)} - file_id: {file_id}")

    print(f"Processing complete. Results saved to {csv_file_path}")
```

```python
csv_file_path = f"./whisper_train.csv"

# Whisper 모델 로드
model = whisper.load_model("base")  # or "medium", "large"
device = "cuda"  # GPU 사용 시

# 특정 폴더만 선택적으로 처리 (중단된 부분부터 재개)
specific_folders = ["[원천]1.AI비서_원천_7_명령어(노인)_training"]

csv_file_path = "./whisper_train.csv"

# 🔥 중단된 부분부터 재개 실행
process_data_resumable(train_path, specific_folders, csv_file_path, model, device)
```

```python
import os

def count_total_wav_files(base_path):
    """
    특정 폴더의 하위 폴더를 순회하며 전체 WAV 파일 개수를 세는 함수.

    Parameters:
        - base_path (str): 기준이 되는 폴더 경로

    Returns:
        - 전체 WAV 파일 개수 (int)
    """
    
    total_wav_count = 0  # 전체 WAV 파일 개수 초기화

    # base_path 내 하위 폴더 가져오기
    subfolders = [f for f in os.listdir(base_path) if os.path.isdir(os.path.join(base_path, f))]

    # 하위 폴더 순회
    for subfolder in subfolders:
        subfolder_path = os.path.join(base_path, subfolder)

        # 현재 하위 폴더 내 WAV 파일 개수 계산
        wav_files = [f for f in os.listdir(subfolder_path) if f.endswith(".wav")]
        wav_count = len(wav_files)

        # 전체 개수 누적
        total_wav_count += wav_count

    print(f"📢 Total .wav files in '{base_path}': {total_wav_count}")
    return total_wav_count

# 사용 예시
base_folder_path = "C:/Users/004/Desktop/data/voice-command/Training/[원천]1.AI비서_원천_8_명령어(노인)_training"
total_files = count_total_wav_files(base_folder_path)
```
